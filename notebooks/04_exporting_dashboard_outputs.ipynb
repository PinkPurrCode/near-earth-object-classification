## Exporting dashboard outputs

```python
# Import pandas for loading and working with the processed NEO dataset
import pandas as pd

# Load the combined dataset created during the data engineering stage
# This file contains the primary asteroid records alongside any matched supplementary orbital data
data = pd.read_csv("combined_neo_dataset.csv")
```

### 1. Dashboard dataset
Create a cleaned record-level dataset for Power BI containing the variables needed for exploratory visuals.

```python
# Select only the variables needed for the Power BI dashboard
# Keeping a separate dashboard dataset avoids exporting unnecessary columns from the full analytical dataset
dashboard_data = data[
    [
        "neo_id",
        "name",
        "absolute_magnitude",
        "estimated_diameter_max",
        "relative_velocity",
        "miss_distance",
        "is_hazardous",
        "e",
        "a",
        "q",
        "i",
        "per",
        "class",
        "matched_supplementary"
    ]
].copy()

# Preview the exported structure to confirm that the expected columns are present
dashboard_data.head()
```

```python
# Check the final number of records and variables before export
# This provides a simple validation that the dashboard dataset has been created at the expected scale
print("Rows:", len(dashboard_data))
print("Columns:", len(dashboard_data.columns))
```

```python
# Export the prepared record-level dataset as a CSV for use in Power BI
# The dataframe index is excluded because it is not a meaningful field for dashboard analysis
dashboard_data.to_csv(
    "neo_dashboard_data.csv",
    index=False
)

# Confirm that the export step has completed
print("neo_dashboard_data.csv created")
```
